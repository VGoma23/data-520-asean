```python
import ee
import geemap
ee.Authenticate()
ee.Initialize(project='iuu-fishing-detections-asean')
```

```python
map = geemap.Map()

eez_boundaries = ee.FeatureCollection('projects/iuu-fishing-detections-asean/assets/eez_boundaries_v12')
map.addLayer(eez_boundaries, {}, 'EEZ boundaries')

map
```

```python
vessel_detections = ee.FeatureCollection('projects/iuu-fishing-detections-asean/assets/sentinel2_vessel_detections_pipev3_20250920')
vessel_detections.first()
```

```python
def get_vessels(feature):
  point = ee.Geometry.Point([feature.get('lon'), feature.get('lat')])
  keep_properties = ['presence_score', 'detect_id', 'matching_score', 'matching_confidence', 'likely_infrastructure', 'potential_ice']

  return ee.Feature(point).copyProperties(feature, keep_properties)

vessel_locations = vessel_detections.map(get_vessels)
```

```python
vessel_locations.first()
```

```python
map.addLayer(vessel_locations, {}, "detection")
map
```

```python
dark_vessels = ee.FeatureCollection('projects/iuu-fishing-detections-asean/assets/ais_disabling_events')
dark_vessels.first()
```

```python
def get_dark_vessels(feature):
  point = ee.Geometry.Point([feature.get('gap_start_lon'), feature.get('gap_start_lat')])
  keep_properties = [ 'gap_id']

  return ee.Feature(point).copyProperties(feature, keep_properties)

dark_locations = dark_vessels.map(get_dark_vessels)
```

```python
dark_locations.first()
```

```python
map.addLayer(dark_locations, {}, "dark vessels")
map
```
